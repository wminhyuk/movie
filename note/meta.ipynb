```python
import pandas as pd
PATH = "~/data/movies/dailyboxoffice/dt=20240101"
df = pd.read_parquet(PATH)
```

```python
df.columns
```

```python
A = pd.DataFrame({
    "movieCd": ["1001", "1002", "1003"],
    "multiMovieYn": ["Y", "Y", "N"],
    "repNationCd": ["K", "F", None]
})
print(A)
```

```python
B = pd.DataFrame({
    "movieCd": ["1001", "1003", "1004"],
    "multiMovieYn": [None, "Y", "Y"],
    "repNationCd": [None, "F", "K"]
})
print(B)
```

```python
A_set = A.set_index('movieCd')
B_set = B.set_index('movieCd')
print("<A_set>", A_set)
print("<B_set>", B_set)
r_df = A_set.combine_first(B_set)
r_df = r_df.reset_index()
print("<r_df>", r_df)
```

```python
merged_df = A.merge(B, on="movieCd", how="outer", suffixes=("_A", "_B"))
merged_df["merged_multiMovieYn"] = merged_df["multiMovieYn_A"].combine_first(merged_df["multiMovieYn_B"])
merged_df["merged_repNationCd"] = merged_df["repNationCd_A"].combine_first(merged_df["repNationCd_B"])
merged_df.drop(columns=["multiMovieYn_A", "multiMovieYn_B", "repNationCd_A", "repNationCd_B"])
```
